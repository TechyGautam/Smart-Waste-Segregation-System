```python
import os

DATASET_PATH = "../dataset"

classes = os.listdir(r"C:\Users\admin\Desktop\waste\final_merged_dataset")

print("Total Classes:", len(classes))
print("\nClass Names:")

for class_name in classes:
    print(class_name)
```

```python
for class_name in classes:

    folder_path = os.path.join(
        r"C:\Users\admin\Desktop\waste\final_merged_dataset",
        class_name
    )

    images = os.listdir(folder_path)

    print(class_name, ":", len(images))
```

```python
import matplotlib.pyplot as plt
from PIL import Image
import os
import random

plt.figure(figsize=(12, 8))

for index, class_name in enumerate(classes):

    folder_path = os.path.join(r"C:\Users\admin\Desktop\waste\final_merged_dataset", class_name)

    image_name = random.choice(os.listdir(folder_path))
    image_path = os.path.join(folder_path, image_name)

    img = Image.open(image_path).convert("RGB")

    plt.subplot(2, 3, index + 1)
    plt.imshow(img)
    plt.title(class_name)
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
for class_name in classes:

    folder_path = os.path.join(r"C:\Users\admin\Desktop\waste\final_merged_dataset", class_name)

    image_name = os.listdir(folder_path)[0]
    image_path = os.path.join(folder_path, image_name)

    img = Image.open(image_path)

    print(class_name, ":", img.size)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = []

for class_name in classes:

    folder_path = os.path.join(r"C:\Users\admin\Desktop\waste\final_merged_dataset", class_name)

    count = len(os.listdir(folder_path))

    data.append({
        "Class": class_name,
        "Count": count
    })

df = pd.DataFrame(data)

plt.figure(figsize=(10, 5))

sns.barplot(
    data=df,
    x="Class",
    y="Count"
)

plt.title("Waste Dataset Distribution")
plt.xticks(rotation=30)
plt.show()
```
